```python
from dotenv import load_dotenv
import os

load_dotenv()
print("Env loaded:", bool(os.getenv("GROQ_API_KEY")))
```

```python
from groq import Groq

client = Groq(api_key=os.getenv("GROQ_API_KEY"))
resp = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": "Say 'CodeGuardian setup verified' and nothing else.",
        }
    ],
)
print(resp.choices[0].message.content)
```

```python
import psycopg2

conn = psycopg2.connect(os.getenv("DATABASE_URL"))
cur = conn.cursor()
cur.execute("SELECT version();")
print(cur.fetchone())
conn.close()
```

```python
from astrapy import DataAPIClient

astra_client = DataAPIClient(os.getenv("ASTRA_DB_APPLICATION_TOKEN"))
db = astra_client.get_database(os.getenv("ASTRA_DB_API_ENDPOINT"))
print("AstraDB collections:", db.list_collection_names())
```

```python
from langfuse import Langfuse

langfuse = Langfuse(
    public_key=os.getenv("LANGFUSE_PUBLIC_KEY"),
    secret_key=os.getenv("LANGFUSE_SECRET_KEY"),
    host=os.getenv("LANGFUSE_HOST"),
)
print("Langfuse auth check:", langfuse.auth_check())
```

```python
from pathlib import Path

from github import Auth, GithubIntegration

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
key_path = PROJECT_ROOT / os.getenv("GITHUB_APP_PRIVATE_KEY_PATH")

with open(key_path, "r") as f:
    private_key = f.read()

auth = Auth.AppAuth(
    app_id=os.getenv("GITHUB_APP_ID"),
    private_key=private_key,
)
integration = GithubIntegration(auth=auth)

installations = integration.get_installations()
for inst in installations:
    print("Installed on:", inst.account.login)
```
